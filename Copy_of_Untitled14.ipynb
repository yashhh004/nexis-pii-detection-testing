```python
import requests

NEXIS_API_KEY = "API"

url = "https://api.nexis.intelation.com/api/text/"
headers = {
    "Authorization": f"Bearer {NEXIS_API_KEY}",
    "Content-Type": "application/json"
}
```

```python
data = {
    "text": "My National Insurance number is PM666111C.",
    "anonymization_method": "mask",
    "country": "United Kingdom",
    "regulation": "UK GDPR"
}

response = requests.post(url, headers=headers, json=data)
print(response.status_code)
print(response.json())
```
